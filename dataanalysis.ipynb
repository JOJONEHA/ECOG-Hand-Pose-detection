```python
non_rest_indices = data[data['paradigm_info'] != 0].index

changes = {}


for channel in range(1, 62):
    pre_mean = data.iloc[non_rest_indices, channel].mean() if len(non_rest_indices) > 0 else 0
    post_mean = data.iloc[data[data['paradigm_info'] == 0].index, channel].mean() if len(data[data['paradigm_info'] == 0].index) > 0 else 0
    changes[f'ECoG_CH{channel}'] = post_mean - pre_mean


changes_df = pd.DataFrame.from_dict(changes, orient='index', columns=['Change'])
top_changes = changes_df.abs().nlargest(15, 'Change')

print("Top 15 Channels with Most Significant Changes (Rest vs Non-Rest):")
print(top_changes)
```

```python
channels = [
    'ECoG_CH21', 'ECoG_CH11', 'ECoG_CH24',
    'ECoG_CH28', 'ECoG_CH29', 'ECoG_CH18',
    'ECoG_CH9', 'ECoG_CH23', 'ECoG_CH36',
    'ECoG_CH17', 'ECoG_CH20', 'ECoG_CH45',
    'ECoG_CH30', 'ECoG_CH7',  'ECoG_CH37'
]

changes = [
    639.77, 505.91, 283.90,
    282.37, 272.18, 267.58,
    265.75, 237.33, 223.53,
    217.79, 164.93, 157.80,
    155.35, 154.05, 147.75
]

colors = plt.cm.viridis(np.linspace(0, 1, len(channels)))

plt.figure(figsize=(12, 8))
plt.barh(channels, changes, color=colors)
plt.xlabel('Change Value')
plt.title('Top 15 ECoG Channels with Significant Changes (Rest vs Non-Rest)')
plt.gca().invert_yaxis()
plt.axvline(0, color='grey', linestyle='--')
plt.show()
```
